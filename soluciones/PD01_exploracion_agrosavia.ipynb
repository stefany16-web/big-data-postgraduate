## 2. Diagnóstico de cinco campos · 10 minutos

Construir una tabla llamada `perfil` para `Secuencial`, `Fecha de Análisis`, `Departamento`, `Municipio` y `pH agua:suelo`. Para cada campo se registran los nulos de pandas, los textos vacíos después de retirar espacios y los marcadores `ND`. Las tres situaciones se cuentan por separado.

📘 **Apoyo:** `isna`, `.str.strip`, `.str.upper`, `.eq` y `sum`.

✅ **Comprobación:** `perfil` debe tener una fila por campo. Con `keep_default_na=False`, un texto vacío o `ND` no se convierte automáticamente en un nulo.

````python
# Desarrollo de la actividad 2
# Crear y mostrar perfil con las cinco columnas solicitadas.

df = pd.read_csv(
    RAW / "agrosavia.csv",
    sep=",",
    encoding="utf-8-sig",
    dtype=str,                 
    keep_default_na=False,     
)

campos = [
    "Secuencial",
    "Fecha de Análisis",
    "Departamento",
    "Municipio",
    "pH agua:suelo"
]

# Construir la tabla perfil
perfil = pd.DataFrame({
    "Campo": campos,
    "Nulos pandas": [
        df[c].isna().sum() for c in campos
    ],
    "Textos vacíos": [
        df[c].str.strip().eq("").sum() for c in campos
    ],
    "Marcadores ND": [
        df[c].str.strip().str.upper().eq("ND").sum()
        for c in campos
    ]
})

perfil
````

**Interpretación del resultado:**

Los resultados muestran que no se encontraron nulos de pandas ni marcadores ND en ninguno de los campos. En Municipio se evidencian dos textos vacíos después de retirar los espacios en blanco. 

## 3. Conversión de pH y fecha · 15 minutos

Crear `ph` desde `pH agua:suelo` y `fecha_dt` desde `Fecha de Análisis`. Utilizar conversión numérica con `errors="coerce"` y fecha con formato `%d/%m/%Y`. Conservar los originales. Distinguir entradas vacías de textos no vacíos que no pudieron convertirse.

📘 **Apoyo:** `pd.to_numeric`, `.str.replace`, `pd.to_datetime`, `isna` y máscaras booleanas.

✅ **Comprobación:** Se muestran los conteos de conversiones fallidas y hasta cinco ejemplos con su valor original. No se reemplazan medidas ausentes por cero.

````python
# Desarrollo de la actividad 3

# Preparar los valores originales para analizarlos
ph_original = df["pH agua:suelo"].astype("string")
fecha_original = df["Fecha de Análisis"].astype("string")

ph_texto = ph_original.str.strip()
fecha_texto = fecha_original.str.strip()

# Convertir el pH a número
ph = pd.to_numeric(
    ph_texto.str.replace(",", ".", regex=False),
    errors="coerce"
)

# Convertir la fecha con el formato día/mes/año
fecha_dt = pd.to_datetime(
    fecha_texto,
    format="%d/%m/%Y",
    errors="coerce"
)

# Agregar las nuevas columnas sin eliminar las originales
df["ph"] = ph
df["fecha_dt"] = fecha_dt

# Identificar entradas vacías
ph_vacio = ph_texto.eq("")
fecha_vacia = fecha_texto.eq("")

# Identificar textos no vacíos que no pudieron convertirse
ph_fallido = ph.isna() & ~ph_vacio
fecha_fallida = fecha_dt.isna() & ~fecha_vacia

# Conteos
print("CONVERSIÓN DEL pH")
print("Entradas vacías:", ph_vacio.sum())
print("Conversiones fallidas no vacías:", ph_fallido.sum())

print("\nCONVERSIÓN DE FECHAS")
print("Entradas vacías:", fecha_vacia.sum())
print("Conversiones fallidas no vacías:", fecha_fallida.sum())

# Cinco ejemplos fallidos con su valor original
print("\nEjemplos de pH que no se pudieron convertir:")
print(
    df.loc[ph_fallido, ["pH agua:suelo"]].head(5)
)

print("\nEjemplos de fechas que no se pudieron convertir:")
print(
    df.loc[fecha_fallida, ["Fecha de Análisis"]].head(5)
)

# Comprobar las columnas nuevas y originales
df[["pH agua:suelo", "ph", "Fecha de Análisis", "fecha_dt"]].head()
````

**Interpretación del resultado:**

pH
- No hay registros sin contenido en la columna del pH.
- Hay 11 valores con contenido que no se pudieron a convertir a número. 

Fechas
- No hay registros sin fecha.
- Todos los valores se convirtieron a fecha válida. 

## 4. Registros para revisión · 15 minutos

Crear la columna booleana `en_revision`. Se marca una fila si su pH es ausente, está fuera de `[0, 14]` o su fecha no se pudo interpretar. Contar por separado cada motivo y el total de filas marcadas. Conservar todas las observaciones.

📘 **Apoyo:** `between`, `isna`, `|`, `~`, `.loc` y `value_counts`.

✅ **Comprobación:** Las filas marcadas y no marcadas deben sumar el total original. Una misma fila puede tener varios motivos; por eso la suma de los motivos no necesariamente coincide con el total marcado.

````python
# Desarrollo de la actividad 4

# Identificar cada motivo por separado

motivo_ph_ausente = df["ph"].isna()

motivo_ph_fuera_rango = ~df["ph"].between(0, 14)

motivo_fecha_invalida = df["fecha_dt"].isna()

# Columna booleana en_revision
df["en_revision"] = (
    motivo_ph_ausente
    | motivo_ph_fuera_rango
    | motivo_fecha_invalida
)

# Filas que cumplen cada motivo
print("pH ausente:", motivo_ph_ausente.sum())

print("pH fuera del rango [0, 14]:",
      motivo_ph_fuera_rango.sum())

print("Fecha no interpretable:",
      motivo_fecha_invalida.sum())

# Total de filas marcadas
print("Total de filas marcadas:",
      df["en_revision"].sum())

# Filas marcadas y no marcadas
print("\nConteo de en_revision:")
print(df["en_revision"].value_counts())

# Comprobar el total de observaciones
print("\nTotal original de observaciones:", len(df))
print("Suma de marcadas y no marcadas:",
      df["en_revision"].sum()
      + (~df["en_revision"]).sum())
````

**Interpretación del resultado:**

Hay 11 filas que deben revisarse porque cumplen al menos una de las condiciones establecidas. 
Hay 92727 filas que no cumplen ninguna de las condiciones y por tanto, no quedan marcadas para revisión. 

## 5. Evidencias y conclusiones · 10 minutos

Construir `controles` con las columnas `control` y `cantidad`: total de entrada, pH sin número, pH fuera del intervalo, fecha no interpretable y total de filas en revisión. Exportar a `OUT / "controles_agrosavia.csv"` sin índice. Redactar dos hallazgos apoyados en los conteos.

📘 **Apoyo:** `pd.DataFrame`, `to_csv(index=False)` y las tablas de las actividades anteriores.

✅ **Comprobación:** El archivo se genera dentro de la ruta `OUT`. Cada hallazgo menciona un conteo y explica su significado o limitación.

````python
# Desarrollo de la actividad 5
# Construir controles, mostrarlo y exportarlo a OUT.

import pandas as pd
from pathlib import Path

# Crear los controles
controles = pd.DataFrame({
    "control": [
        "Total de entrada",
        "pH sin número",
        "pH fuera del intervalo",
        "Fecha no interpretable",
        "Total de filas en revisión"
    ],
    "cantidad": [
        len(df),
        df["ph"].isna().sum(),
        (df["ph"].notna() & ~df["ph"].between(0, 14)).sum(),
        df["fecha_dt"].isna().sum(),
        df["en_revision"].sum()
    ]
})

print(controles)

# Crear la carpeta OUT si no existe
OUT = Path("OUT")
OUT.mkdir(parents=True, exist_ok=True)

# Exportar sin índice
controles.to_csv(OUT / "controles_agrosavia.csv", index=False)

print("\nArchivo guardado en:", OUT / "controles_agrosavia.csv")
````

**Hallazgos**

1. **Control de pH:** De los 92.738 registros de entrada, se identificaron 11 registros con pH sin número y ninguno con valores fuera del intervalo permitido de 0 a 14. Esto indica que los problemas relacionados con el pH corresponden a valores ausentes o no numéricos, por lo que deben revisarse antes de realizar los análisis.

2. **Control de fechas y revisión:** Se encontraron 0 fechas no interpretables y 11 filas marcadas para revisión. Esto significa que todas las filas que requieren revisión están relacionadas con el pH sin número, mientras que las fechas pudieron interpretarse correctamente. Por lo tanto, es necesario verificar esos 11 registros antes de utilizar los datos en los análisis.
